### Accès au fichier par son chemin
Une fois le Drive monté, vous pouvez définir le chemin de votre fichier (par exemple, un fichier CSV ou texte) et l'ouvrir. Remplacez `'MonDossier/mon_fichier.csv'` par le chemin réel de votre fichier dans votre Drive.

```python
import os
import pandas as pd

# Définissez le chemin d'accès relatif à votre dossier 'My Drive'
chemin_drive = 'MonDossier/mon_fichier.csv'

# Chemin d'accès complet dans l'environnement Colab
chemin_complet = os.path.join('/content/drive/MyDrive', chemin_drive)

print(f"Recherche du fichier à l'adresse : {chemin_complet}")

# Exemple d'ouverture si c'est un fichier CSV (à adapter selon votre format)
try:
    if os.path.exists(chemin_complet):
        # Si c'est un fichier CSV
        df = pd.read_csv(chemin_complet)
        display(df.head())
        # Si vous souhaitez simplement lire un fichier texte, décommentez les lignes suivantes :
        # with open(chemin_complet, 'r') as f:
        #     print(f.read())
    else:
        print("Le fichier n'a pas été trouvé. Veuillez vérifier le chemin d'accès.")
except Exception as e:
    print(f"Une erreur est survenue lors de l'ouverture : {e}")
```
